# Analysis & Visualisations

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data/retail_clean.csv')

# Monthly Sales Trend
monthly = (df.groupby(['Year', 'Month'])['TotalPrice']
             .sum()
             .reset_index()
             .sort_values(['Year', 'Month']))
monthly['YearMonth'] = monthly['Year'].astype(str) + '-' + monthly['Month'].astype(str).str.zfill(2)

plt.figure(figsize=(12,6))
plt.plot(monthly['YearMonth'], monthly['TotalPrice'], marker='o')
plt.title('Monthly Sales Trend')
plt.xticks(rotation=60)
plt.ylabel('Sales (£)')
plt.grid(True)
plt.show()
```

```python
# Top 10 Products
top_prod = (df.groupby('Description')['TotalPrice']
              .sum()
              .sort_values(ascending=False)
              .head(10))

plt.figure(figsize=(10,6))
top_prod.sort_values().plot(kind='barh')
plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales (£)')
plt.tight_layout()
plt.show()
```

```python
# Top 10 Countries
top_country = (df.groupby('Country')['TotalPrice']
                 .sum()
                 .sort_values(ascending=False)
                 .head(10))
plt.figure(figsize=(8,6))
top_country.plot(kind='bar')
plt.title('Top 10 Countries by Sales')
plt.ylabel('Total Sales (£)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
