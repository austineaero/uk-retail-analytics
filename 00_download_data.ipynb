# Download & Prepare Data
Fetch the UCI Online Retail Excel file, convert to CSV, and save into the `data/` directory.

```python
pip install -q -r requirements.txt
```

```python
import os
import pandas as pd
import requests
import io

os.makedirs('data', exist_ok=True)  # creates the data folder if it doesn't exist

url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/00352/Online%20Retail.xlsx'
print('Downloading dataset...')
resp = requests.get(url, timeout=60)
resp.raise_for_status()

# Read directly from the Excel bytes
df = pd.read_excel(io.BytesIO(resp.content))
csv_path = 'data/online_retail.csv'
df.to_csv(csv_path, index=False)
print(f'Saved CSV to {csv_path}')
df.head()
```
