# Export Aggregates for Dashboarding

```python
import pandas as pd

# Read cleaned retail data
df = pd.read_csv('data/retail_clean.csv')

# Monthly Sales Export
monthly_sales = (df.groupby(['Year', 'Month'])['TotalPrice']
                   .sum()
                   .reset_index()
                   .rename(columns={'TotalPrice':'MonthlySales'}))
monthly_sales.to_csv('data/monthly_sales.csv', index=False)

# Top 20 Products Export
top_products = (df.groupby('Description')['TotalPrice']
                  .sum()
                  .sort_values(ascending=False)
                  .head(20)
                  .reset_index())
top_products.to_csv('data/top_products.csv', index=False)

# Top 20 Countries Export
top_countries = (df.groupby('Country')['TotalPrice']
                   .sum()
                   .sort_values(ascending=False)
                   .head(20)
                   .reset_index())
top_countries.to_csv('data/top_countries.csv', index=False)

# Print summary of what was exported
for filename in ['monthly_sales.csv', 'top_products.csv', 'top_countries.csv']:
    print(f" - data/{filename}")
```
